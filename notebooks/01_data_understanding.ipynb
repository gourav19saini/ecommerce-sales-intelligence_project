## Initial observations

- The dataset contains **9,994 transaction-level rows and 21 columns**, covering US Superstore orders from 2014-2017.
- Core numerical fields are `Sales`, `Quantity`, `Discount`, and `Profit`.
- `Order Date` and `Ship Date` are stored as text and need conversion to datetime (handled in the next notebook).
- Categorical fields (`Segment`, `Region`, `Category`, `Sub-Category`, `Ship Mode`) have a small number of unique values and are good candidates for `category` dtype.
- No missing values or duplicate rows were detected in this dataset, so the cleaning notebook focuses mainly on type conversion and outlier inspection rather than imputation.
- `Customer ID` and `Product ID` allow customer- and product-level feature engineering later in the pipeline.
